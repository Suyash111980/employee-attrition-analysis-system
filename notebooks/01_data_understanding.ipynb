```python
# Load the cleaned dataset for EDA

from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent

PROCESSED_DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "employee_attrition_cleaned.csv"
)

df_eda = pd.read_csv(PROCESSED_DATA_PATH)

print("Cleaned dataset loaded successfully.")
print("Shape:", df_eda.shape)
```

```python
# Attrition by Department

department_attrition = pd.crosstab(
    df_eda["Department"],
    df_eda["Attrition"],
    normalize="index"
) * 100

department_attrition.round(2)
```

```python
# Visualize attrition percentage by department

department_attrition.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Attrition Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Percentage of Employees")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()
```

```python
# Attrition by OverTime

overtime_attrition = pd.crosstab(
    df_eda["OverTime"],
    df_eda["Attrition"],
    normalize="index"
) * 100

overtime_attrition.round(2)
```

```python
# Visualize attrition percentage by OverTime

overtime_attrition.plot(
    kind="bar",
    figsize=(7, 5)
)

plt.title("Attrition Distribution by OverTime")
plt.xlabel("OverTime")
plt.ylabel("Percentage of Employees")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()
```

```python
# Visualize attrition percentage by Job Role

jobrole_attrition.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Attrition Distribution by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Percentage of Employees")
plt.xticks(rotation=45, ha="right")
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()
```

```python
# Create age groups

age_bins = [17, 25, 35, 45, 55, 100]
age_labels = ["18-25", "26-35", "36-45", "46-55", "56+"]

df_eda["AgeGroup"] = pd.cut(
    df_eda["Age"],
    bins=age_bins,
    labels=age_labels
)

# Calculate attrition percentage by age group

age_attrition = pd.crosstab(
    df_eda["AgeGroup"],
    df_eda["Attrition"],
    normalize="index"
) * 100

age_attrition.round(2)
```

```python
# Visualize attrition percentage by age group

age_attrition.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Attrition Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Percentage of Employees")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()
```

```python
# Select numerical variables
numeric_df = df_eda.select_dtypes(include=["int64", "float64"])

# Calculate correlation matrix
correlation_matrix = numeric_df.corr()

correlation_matrix.round(2)
```

```python
# Correlation heatmap

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    annot=False,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
# Find the strongest correlations between different numerical variables

correlation_pairs = (
    correlation_matrix
    .where(~np.eye(correlation_matrix.shape[0], dtype=bool))
    .stack()
    .reset_index()
)

correlation_pairs.columns = ["Variable_1", "Variable_2", "Correlation"]

correlation_pairs["AbsCorrelation"] = (
    correlation_pairs["Correlation"].abs()
)

strongest_correlations = (
    correlation_pairs
    .sort_values("AbsCorrelation", ascending=False)
    .drop_duplicates(subset=["AbsCorrelation"])
    .head(15)
)

strongest_correlations[
    ["Variable_1", "Variable_2", "Correlation"]
].round(3)
```
